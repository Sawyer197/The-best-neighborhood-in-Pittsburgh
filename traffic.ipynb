```python
import pandas as pd

traffic = pd.read_csv("https://data.wprdc.org/datastore/dump/6dfd4f8f-cbf5-4917-a5eb-fd07f4403167")
traffic.head/()
```

```python
traffic[['neighborhood', 'average_daily_car_traffic']].dropna().head(10)
```

```python
traffic.groupby('neighborhood')['average_daily_car_traffic'].mean().sort_values().head(10)
```

I used the city traffic dataset and focused on average daily car traffic by neighborhood. Then I compared the average traffic values across neighborhoods. Lower traffic may suggest a quieter neighborhood.

```python
traffic_by_neighborhood = traffic.groupby('neighborhood')['average_daily_car_traffic'].mean().sort_values().head(10)

traffic_by_neighborhood.plot(kind='barh', figsize=(10,6), title='Lowest Average Daily Car Traffic by Neighborhood')
```
